```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("--- VERIFICA LIVE: INTERROGAZIONE API openFDA FAERS ---")

# Endpoint ufficiale openFDA per gli eventi avversi
faers_endpoint = (
    'https://api.fda.gov/drug/event.json?'
    'search=patient.drug.openfda.generic_name:"pioglitazone"+AND+'
    '('
    'patient.reaction.reactionmeddrapt.exact:"Exposure during pregnancy"+OR+'
    'patient.reaction.reactionmeddrapt.exact:"Maternal exposure during pregnancy"+OR+'
    'patient.reaction.reactionmeddrapt.exact:"Pregnancy"+OR+'
    'patient.reaction.reactionmeddrapt.exact:"Congenital anomaly"+OR+'
    'patient.reaction.reactionmeddrapt.exact:"Spontaneous abortion"+OR+'
    'patient.reaction.reactionmeddrapt.exact:"Foetal growth restriction"'
    ')'
    '&count=patient.reaction.reactionmeddrapt.exact'
)

print("Invio richiesta HTTP in tempo reale ai server FDA...")
try:
    response = requests.get(faers_endpoint, timeout=15)
    
    if response.ok:
        json_data = response.json()
        results = json_data.get('results', [])
        
        df_live_faers = pd.DataFrame(results)
        df_live_faers['Term_Clean'] = df_live_faers['term'].str.capitalize()
        df_live_faers = df_live_faers.sort_values(by='count', ascending=True).reset_index(drop=True)
        
        print(f"\n✔ QUERY RIUSCITA! Trovate {df_live_faers['count'].sum():,} segnalazioni totali di gravidanza nel database FAERS.")
        print("\nDettaglio per termine MedDRA (Preferred Term):")
        display(df_live_faers[['Term_Clean', 'count']].iloc[::-1])
        
        # Generazione rapida del grafico di controllo
        fig, ax = plt.subplots(figsize=(10, 5))
        bars = ax.barh(df_live_faers['Term_Clean'], df_live_faers['count'], color='#1565c0', height=0.6, edgecolor='#263238')
        ax.set_xlabel("Spontaneous Reports Count (FAERS Database)", fontsize=10.5, fontweight='bold')
        ax.set_title("Live openFDA FAERS Query: Pioglitazone & Pregnancy-Related Adverse Events", fontsize=11.5, fontweight='bold', pad=12)
        ax.grid(axis='x', linestyle=':', alpha=0.6)
        
        for bar in bars:
            w = bar.get_width()
            ax.text(w + 0.3, bar.get_y() + bar.get_height()/2, f"{int(w)}", va='center', ha='left', fontsize=9.5, fontweight='bold', color='#0d47a1')
            
        plt.tight_layout()
        plt.show()
        print("\n✔ Controllo FAERS completato con successo!")
        
    else:
        print(f"❌ Errore HTTP dal server FDA: {response.status_code}")
        print(response.text)

except Exception as e:
    print(f"❌ Errore di connessione o timeout: {e}")
```
